### 1. Algebra Review 

- In Data Science, Logarithms are important for compressing highly skewed data like income or web traffic.
- Real World Example : Transforming a highly skewed dataset of real estate prices so a linear regression model can predict home values accurately without being distorted by multi millionaire dollar outliers.

```python
# - Python Implementation

import numpy as np
import pandas as pd

# Creating a highly skewed dataset (e.g., Household Income)
incomes = np.random.exponential(scale=50000, size=1000) + 15000

# Applying log transformation to stabilize variance
log_incomes = np.log(incomes)

print(f"Original Skew: {pd.Series(incomes).skew():.2f}")
print(f"Log-Transformed Skew: {pd.Series(log_incomes).skew():.2f}")
```

## 2. Sigma Notation ($\sum$) and Basic Calculus

- **Why it matters:** Sigma notation is the mathematical shorthand for loops. Calculus (specifically derivatives) is the engine behind Gradient Descent, which minimizes error in models like Neural Networks and Logistic Regressions.

- **Intuition:** A derivative measures the instantaneous rate of change—the exact slope of a curve at a single point. If you are rolling down a hill, the derivative tells you how fast your speed is changing at that exact millisecond.

- Mathematical Explanation:The sum of $n$ data points:$$\sum_{i=1}^{n} x_i = x_1 + x_2 + \dots + x_n$$The derivative of a cost function $J(w) = w^2$ with respect to weight $w$ tells us how to update $w$ to find the minimum:$$\frac{d}{dw}(w^2) = 2w$$

```python
# Python Implementation:

# Sigma Notation implemented via Vectorization (No explicit for-loops)
x = np.array([2, 4, 6, 8, 10])
sum_x = np.sum(x)  # Replaces mathematical Sigma notation

# Basic numeric derivative evaluation (Gradient estimation)
def cost_function(w):
    return w ** 2

def compute_gradient(w, h=1e-5):
    return (cost_function(w + h) - cost_function(w)) / h

print(f"Sum: {sum_x} | Estimated Gradient at w=3: {compute_gradient(3):.4f}")
```

## 1. Central Tendency & Dispersion

- **Why it matters** - Mean, median, and mode pinpoint where your data aggregates. Varianceand standard deviation describe how scattered your data points are around that center.
- **Intuition** - The mean tells you the center; variance tells you the risk.
