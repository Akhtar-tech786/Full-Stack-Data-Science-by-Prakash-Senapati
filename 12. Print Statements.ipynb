#### print Format method

```python
num1=20 
num2=30 
add=num1+num2
print('The addition of {} and {} is = {}'.format(num1,num2,add))
```

- first deside how the print statement should be
- like:- The addition of 20 and 30 is = 50
- then replace the variable position with curly bracets { }
- then appply.format (val1,val2,....val - method

```python
name='Python'
age=20
city='hyd'
# hello my name is python and iam 10 year old from hydrabad
```

```python
print('hellomynameis{},andiam{}yearsoldfrom{}'.format(name,age,city))
```

```python
num1=100
num2=25
num3=333
avg=(num1+num2+num3)/3  # or we can use avg = round(num1+num2+num3)/3,2) 
avg1=round((num1+num2+num3)/3,2) 
#The avrage of num1,num2,num3 is = avg 
print('The average of {}, {}, and {} is = {} or {}'.format(num1,num2,num3,avg,avg1)) #here we can use round(avg,2) also 
```

```python
round(avg,2) #round of till 2digit after decimal
```
